# 01 — Eksplorativna analiza
Pregled sintetičkog Rossmann-like dataset-a. Cilj: razumjeti sezonske obrasce, uticaj promocija i praznika prije modeliranja.

```python
import sys, pathlib
ROOT = pathlib.Path('..').resolve()
sys.path.insert(0, str(ROOT))

import pandas as pd
import matplotlib.pyplot as plt

from src.generate_sample_data import ensure_sample_data
from src.data_processing import load_sales_csv, basic_summary, filter_series

DATA = ROOT / 'data' / 'sample_sales.csv'
ensure_sample_data(DATA)
df = load_sales_csv(DATA)
basic_summary(df)
```

```python
df.head()
```

## Prosječna prodaja po danu u sedmici

```python
df['dow'] = df['Date'].dt.dayofweek
df.groupby('dow')['Sales'].mean().plot(kind='bar', title='Prosječna prodaja po danu u sedmici')
plt.xlabel('Dan (0=pon, 6=ned)'); plt.ylabel('Avg sales'); plt.show()
```

## Sezonski obrazac (mjesečni nivo)

```python
df['month'] = df['Date'].dt.month
df.groupby('month')['Sales'].mean().plot(marker='o', title='Sezonski prosjek po mjesecu')
plt.xlabel('Mjesec'); plt.ylabel('Avg sales'); plt.show()
```

## Uticaj promocije

```python
if 'Promo' in df.columns:
    df.groupby('Promo')['Sales'].mean().plot(kind='bar', title='Prosječna prodaja — Promo da/ne')
    plt.show()
```

## Vremenska serija za jednu prodavnicu/proizvod

```python
series = filter_series(df, df['Store'].iloc[0], df['Product'].iloc[0])
series.set_index('Date')['Sales'].plot(figsize=(12,4), title='Dnevna prodaja')
plt.show()
```
